```python
# cmap and alpha(0 to 1)
```

- cmap is a Matplotlib parameter that applies a color map to numerical data, representing different values with different colors

- alpha controls the transparency of plot elements.
- alpha = 0 → Fully transparent (invisible)
- alpha = 1 → Fully opaque (default)

### 3D Scatter Plots

A **3D Scatter Plot** is used to display the relationship between **three numerical variables**. Each point is represented by three coordinates:

- **X-axis** → First variable
- **Y-axis** → Second variable
- **Z-axis** → Third variable

It is commonly used in **Machine Learning**, **Data Science**, and **scientific visualization**.


```python
batters
fig =  plt.figure()

ax = plt.subplot(projection='3d')
```

```python
batters

fig = plt.figure()

ax = plt.subplot(projection='3d')

ax.scatter3D(batters['runs'],batters['avg'],batters['strike_rate'],marker='+')
ax.set_title('IPL batsman analysis')

ax.set_xlabel('Runs')
ax.set_ylabel('Avg')
ax.set_zlabel('SR')
```

### 3D Line Plot

A **3D Line Plot** is used to display the relationship between **three numerical variables** by connecting data points with a continuous line in three-dimensional space.

- **X-axis** → First variable
- **Y-axis** → Second variable
- **Z-axis** → Third variable

It is commonly used to visualize **trends**, **trajectories**, **time-series data**, and **scientific or engineering data** in three dimensions.

```python
x = [0,1,5,25]
y = [0,10,13,0]
z = [0,13,20,9]

fig = plt.figure()

ax = plt.subplot(projection='3d')

ax.scatter3D(x,y,z,s=[100,100,100,100])
ax.plot3D(x,y,z,color='red')
```

### 3D Surface Plots

A **3D Surface Plot** is used to visualize a **continuous three-dimensional surface**. It represents the relationship between **two independent variables (X and Y)** and **one dependent variable (Z)**.

- **X-axis** → First independent variable
- **Y-axis** → Second independent variable
- **Z-axis** → Dependent variable (surface height)

It is commonly used in **Machine Learning**, **Data Science**, **engineering**, **scientific visualization**, and **mathematical function analysis**.<br>
### Uses of 3D Surface Plot

- Visualize continuous 3D surfaces.
- Visualize **loss (cost) functions** in Machine Learning.

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx, yy = np.meshgrid(x,y)
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.log(xx)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

### Contour Plots

## Definition
A **Contour Plot** is a **2D representation of a 3D surface**. It displays **lines (contours)** that connect points having the **same Z-value (height)**.

## Uses
- Visualize a **3D surface in 2D**.
- Visualize **loss (cost) functions** in Machine Learning.
- Analyze mathematical functions.
- Display terrain and elevation maps.
- Used in **Machine Learning**, **Data Science**, **engineering**, and **scientific visualization**.


```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
# contourf

fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
# contourf -> color is the 3rd axis to know top and bottom
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

### Heatmap

#### Definition
A **Heatmap** is a graphical representation of data in which **values are represented by different colors**. Higher and lower values are shown using different color intensities, making patterns easy to identify.

#### Uses
- Visualize the distribution of data.
- Display correlation matrices.
- Identify patterns, trends, and outliers.
- Analyze large datasets.
- Used in **Machine Learning**, **Data Science**, **statistics**, and **scientific visualization**.

```python
delivery = pd.read_csv('IPL_Ball_by_Ball_2008_2022.csv')
delivery.head()
```

```python
temp_df = delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run']==6)]
```

```python
grid = temp_df.pivot_table(index='overs',columns='ballnumber',values='batsman_run',aggfunc='count')
```

```python
plt.figure(figsize=(20,10))
plt.imshow(grid)
plt.yticks(delivery['overs'].unique(), list(range(1,21)))
plt.xticks(np.arange(0,6), list(range(1,7)))
plt.colorbar()
```
